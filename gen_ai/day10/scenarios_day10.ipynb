```python
# 🧠 AutoGen Assessment Scenario
# 🎯 Title: Multi-Agent AI System for Automated Code Review & Execution
# 🔹 Scenario Overview

# You are part of an AI engineering team building an AutoGen-based developer assistant.

# The system should:

# Accept a user problem statement
# Generate code
# Review the code
# Execute it
# Provide feedback and corrections

# This must be implemented using multiple collaborating agents.

# 🔹 Problem Statement

# A user gives the following input:

# “Write a Python function to find the second largest number in a list and handle edge cases.”

# Your AutoGen system should:

# Generate correct code
# Validate logic
# Execute the code
# Handle errors
# Provide final response
# 🔹 Requirements
# ✅ 1. Agent Design (MANDATORY)

# Create at least 3 agents:

# 🔸 Coder Agent
# Generates Python code
# Should follow best practices
# Must handle edge cases
# 🔸 Reviewer Agent
# Reviews code for:
# Bugs
# Logic errors
# Edge cases
# Suggests improvements
# 🔸 Executor Agent (UserProxy)
# Runs the code
# Returns output/errors
# 🔹 2. Agent Collaboration Flow

# Design the flow such that:

# User → Coder → Reviewer → Executor → Reviewer (if error) → Final Output

# 👉 The system should not stop after one step
# 👉 It must iterate until correct output is achieved

# 🔹 3. Functional Expectations

# Your system must:

# ✔ Handle incorrect code generated initially
# ✔ Ensure reviewer gives structured feedback
# ✔ Prevent infinite loops
# ✔ Show final clean output

# 🔹 4. Technical Constraints
# Use AutoGen framework
# Use LLM-based agents
# Enable code execution
# Add termination condition
# 🔹 5. Edge Cases to Handle

# Test your system with:

# Empty list
# List with one element
# Duplicate values
# Negative numbers


# 1. Setup & API Configuration
 
import os
from dotenv import load_dotenv
import autogen

load_dotenv()

api_key = os.getenv("GROQ_API_KEY")
if not api_key:
    raise ValueError("GROQ_API_KEY not found")

os.environ["GROQ_API_KEY"] = api_key

config_list = [
    {
        "model": "llama-3.1-8b-instant",  # ✅ stable + low token
        "api_key": os.environ["GROQ_API_KEY"],
        "base_url": "https://api.groq.com/openai/v1",
    }
]

llm_config = {
    "config_list": config_list,
    "temperature": 0.3,
    "max_tokens": 500
}


# ==============================
# 2. Define Agents
# ==============================

# 🔹 Coder Agent
coder = autogen.AssistantAgent(
    name="Coder",
    llm_config=llm_config,
    max_consecutive_auto_reply=2,
    system_message="""
You are a Python expert.

Task:
- Write clean, correct Python code
- Solve the problem: find second largest number in a list
- Handle edge cases:
    * Empty list
    * One element
    * Duplicate values
    * Negative numbers
- Return ONLY code (no explanation)
"""
)


# 🔹 Reviewer Agent
reviewer = autogen.AssistantAgent(
    name="Reviewer",
    llm_config=llm_config,
    system_message="""
You are a strict code reviewer.

Responsibilities:
- Check for bugs and logic errors
- Validate edge case handling
- Ensure Python best practices
- If code is correct → say "APPROVED"
- If incorrect → clearly explain issues and suggest fixes
"""
)


# 🔹 Executor Agent
executor = autogen.UserProxyAgent(
    name="Executor",
    human_input_mode="NEVER",
    code_execution_config={
        "work_dir": "coding",
        "use_docker": False
    }
)


# ==============================
# 3. Multi-Agent Workflow Logic
# ==============================

def run_autogen_system(problem):

    print("\n🚀 Starting System...\n")

    coder_response = coder.generate_reply(
        messages=[{"role": "user", "content": str(problem)}]
    )

    coder_reply = coder_response if isinstance(coder_response, str) else str(coder_response)

    max_iterations = 5

    for i in range(max_iterations):

        print(f"\n🔍 Iteration {i+1}\n")

        if not coder_reply.strip():
            coder_reply = "No code generated. Please regenerate."

        review = reviewer.generate_reply(
            messages=[{"role": "user", "content": str(coder_reply)}]
        )

        review = review if isinstance(review, str) else str(review)

        print("🧠 Reviewer:\n", review)

        if "APPROVED" in review:

            print("\n✅ Approved → Executing...\n")

            try:
                exec_result = executor.run_code(coder_reply)
                print("⚙️ Result:\n", exec_result)
            except Exception as e:
                print("❌ Execution Error:", e)

            return coder_reply

        else:
            print("\n🔁 Fixing code...\n")

            coder_response = coder.generate_reply(
                messages=[
                    {"role": "user", "content": str(problem)},
                    {"role": "assistant", "content": str(coder_reply)},
                    {"role": "user", "content": str(review)}
                ]
            )

            coder_reply = coder_response if isinstance(coder_response, str) else str(coder_response)

    print("\n⚠️ Max iterations reached.\n")
    return coder_reply

# ==============================
# 4. Run the System
# ==============================

problem_statement = """
Write a Python function to find the second largest number in a list.
Handle all edge cases and include test examples.
"""

final_code = run_autogen_system(problem_statement)

print("\n🎯 Final Code:\n", final_code)
```
