```python
# Scenario: AI Content Automation for a Startup
# 🏢 Problem Statement

# A startup wants to:

# Stay updated with latest AI trends

# Automatically create LinkedIn posts

# Save time on manual research & writing

# 👉 So we build a multi-agent AI system to automate this using CrewAI.
```

```python
!pip install crewai litellm langchain-community langchain-litellm

import os
os.environ["CREWAI_TRACING_ENABLED"] = "true"
from crewai import Agent, Task, Crew
from google.colab import userdata
from langchain_litellm import ChatLiteLLM # Updated import

# -----------------------------
# 1. Set Groq API Key
# -----------------------------
os.environ["GROQ_API_KEY"] = userdata.get("GROQ_API_KEY")

# ❌ REMOVE any OPENAI key completely
if "OPENAI_API_KEY" in os.environ:
    del os.environ["OPENAI_API_KEY"]

# -----------------------------
# 2. Define LLM (using ChatLiteLLM for Groq)
# -----------------------------
llm = ChatLiteLLM(model="groq/llama-3.3-70b-versatile")

# -----------------------------
# 3. Define Agents
# -----------------------------
researcher = Agent(
    role="Researcher",
    goal="Find latest AI news",
    backstory="Expert in AI trends",
    llm=llm, # Use the explicitly defined llm object
    verbose=True
)

writer = Agent(
    role="Writer",
    goal="Write LinkedIn post",
    backstory="Expert content creator",
    llm=llm, # Use the explicitly defined llm object
    verbose=True
)

# -----------------------------
# 4. Tasks
# -----------------------------
research_task = Task(
    description="Find 5 latest AI news trends",
    expected_output="List with summaries",
    agent=researcher
)

write_task = Task(
    description="Write LinkedIn post",
    expected_output="Final post",
    agent=writer,
    context=[research_task]
)

# -----------------------------
# 5. Crew
# -----------------------------
crew = Crew(
    agents=[researcher, writer],
    tasks=[research_task, write_task],
    verbose=True
)

# -----------------------------
# 6. Run
# -----------------------------
result = crew.kickoff()

print("\n✅ FINAL OUTPUT:\n")
print(result)
```

```python
# Scenario: Automated Learning Hub for an EdTech Company
# 🎓 Problem Statement
# An EdTech company wants to:
# - Keep track of the latest educational technology innovations
# - Automatically generate weekly blog posts for their website
# - Reduce time spent on manual research and content drafting
# 👉 So we build a multi-agent AI system using CrewAI to automate this workflow:
# - Trend Monitoring Agent → Continuously scans for new EdTech tools, teaching strategies, and AI-in-education updates
# - Content Summarization Agent → Condenses research into digestible insights tailored for educators and learners
# - Blog Writing Agent → Crafts engaging, SEO-friendly blog posts automatically
# - Publishing Agent → Schedules and posts content directly to the company’s website


# Install (run once)
# !pip install crewai litellm langchain-community langchain-litellm python-dotenv

import os
from dotenv import load_dotenv

# -----------------------------
# 1. Load API Key
# -----------------------------
load_dotenv()

api_key = os.getenv("GROQ_API_KEY")
if not api_key:
    raise ValueError("❌ GROQ_API_KEY not found")

os.environ["GROQ_API_KEY"] = api_key

# -----------------------------
# 2. Imports
# -----------------------------
from crewai import Agent, Task, Crew
from langchain_litellm import ChatLiteLLM

# -----------------------------
# 3. Define LLM
# -----------------------------
llm = ChatLiteLLM(
    model="groq/llama-3.3-70b-versatile",
    temperature=0.7
)

# -----------------------------
# 4. Define Agents (UPDATED 🔥)
# -----------------------------

# 🔍 Trend Monitoring Agent
trend_agent = Agent(
    role="Trend Monitoring Agent",
    goal="Find latest trends in EdTech, AI in education, and learning tools",
    backstory="Expert in tracking global education technology innovations",
    llm=llm,
    verbose=True
)

# 🧠 Content Summarization Agent
summary_agent = Agent(
    role="Content Summarization Agent",
    goal="Summarize complex EdTech trends into simple insights",
    backstory="Expert in simplifying research for educators and students",
    llm=llm,
    verbose=True
)

# ✍️ Blog Writing Agent
writer_agent = Agent(
    role="Blog Writing Agent",
    goal="Write engaging, SEO-friendly blog posts",
    backstory="Professional blog writer specializing in EdTech content",
    llm=llm,
    verbose=True
)

# 🚀 Publishing Agent
publisher_agent = Agent(
    role="Publishing Agent",
    goal="Prepare and schedule blog post for publishing",
    backstory="Content manager responsible for formatting and publishing blogs",
    llm=llm,
    verbose=True
)

# -----------------------------
# 5. Define Tasks (UPDATED 🔥)
# -----------------------------

# Step 1: Research trends
trend_task = Task(
    description="""
    Find 5 latest trends in EdTech including:
    - AI in education
    - Online learning tools
    - Innovative teaching methods
    Include short explanations for each.
    """,
    expected_output="List of 5 EdTech trends with explanations",
    agent=trend_agent
)

# Step 2: Summarize
summary_task = Task(
    description="""
    Convert the trends into simple, easy-to-understand insights
    suitable for students and educators.
    """,
    expected_output="Simplified insights from trends",
    agent=summary_agent,
    context=[trend_task]   # 🔥 shared memory
)

# Step 3: Write blog
blog_task = Task(
    description="""
    Write a complete blog post based on the summarized insights.
    Include:
    - Catchy title
    - Introduction
    - Main content (structured)
    - Conclusion
    - SEO-friendly tone
    """,
    expected_output="Full blog post",
    agent=writer_agent,
    context=[summary_task]
)

# Step 4: Publish
publish_task = Task(
    description="""
    Format the blog post for publishing.
    Add:
    - Proper headings
    - Bullet points
    - Final polished version ready to post
    """,
    expected_output="Final formatted blog ready for publishing",
    agent=publisher_agent,
    context=[blog_task]
)

# -----------------------------
# 6. Create Crew
# -----------------------------
crew = Crew(
    agents=[trend_agent, summary_agent, writer_agent, publisher_agent],
    tasks=[trend_task, summary_task, blog_task, publish_task],
    verbose=True
)

# -----------------------------
# 7. Run Workflow
# -----------------------------
result = crew.kickoff()

print("\n✅ FINAL BLOG OUTPUT:\n")
print(result)
```

```python
import os
from dotenv import load_dotenv

# -----------------------------
# 1. Load API Key
# -----------------------------
load_dotenv()

api_key = os.getenv("GROQ_API_KEY")
if not api_key:
    raise ValueError("❌ GROQ_API_KEY not found")

os.environ["GROQ_API_KEY"] = api_key

# -----------------------------
# 2. Imports
# -----------------------------
from crewai import Agent, Task, Crew
from langchain_litellm import ChatLiteLLM

# -----------------------------
# 3. Define LLM
# -----------------------------
llm = ChatLiteLLM(
    model="groq/llama-3.3-70b-versatile",
    temperature=0.7
)

# -----------------------------
# 4. Define Agents (SALES SYSTEM 🔥)
# -----------------------------

# 📄 Agent 1: RFP Analyzer
rfp_agent = Agent(
    role="RFP Analyzer",
    goal="Extract key requirements, budget, and evaluation criteria from RFP documents",
    backstory="Expert business analyst skilled in interpreting enterprise RFPs",
    llm=llm,
    verbose=True
)

# 🧩 Agent 2: Solution Mapper
solution_agent = Agent(
    role="Solution Mapper",
    goal="Map company offerings to client requirements and suggest value propositions",
    backstory="Senior solutions architect with deep knowledge of products and services",
    llm=llm,
    verbose=True
)

# ✍️ Agent 3: Proposal Writer
writer_agent = Agent(
    role="Proposal Writer",
    goal="Generate a professional, persuasive sales proposal tailored to the client",
    backstory="Expert copywriter specializing in enterprise sales proposals",
    llm=llm,
    verbose=True
)

# -----------------------------
# 5. Define Tasks (WORKFLOW 🔄)
# -----------------------------

# Step 1: Analyze RFP
rfp_task = Task(
    description="""
    Analyze the given RFP document and extract:
    - Key business requirements
    - Budget constraints (if mentioned)
    - Evaluation criteria
    - Pain points of the client
    """,
    expected_output="Structured summary of RFP requirements",
    agent=rfp_agent
)

# Step 2: Map Solution
solution_task = Task(
    description="""
    Based on the RFP analysis:
    - Map company products/services to each requirement
    - Suggest value propositions
    - Highlight differentiators vs competitors
    """,
    expected_output="Mapped solutions with value propositions",
    agent=solution_agent,
    context=[rfp_task]   # 🔥 shared memory
)

# Step 3: Write Proposal
proposal_task = Task(
    description="""
    Create a complete sales proposal including:
    - Executive Summary
    - Understanding of Client Needs
    - Proposed Solution
    - Benefits & Differentiators
    - Pricing Overview (if possible)
    - Conclusion

    Ensure:
    - Professional tone
    - Client-specific customization
    - Persuasive writing
    """,
    expected_output="Full client-ready sales proposal",
    agent=writer_agent,
    context=[solution_task]
)

# -----------------------------
# 6. Create Crew
# -----------------------------
crew = Crew(
    agents=[rfp_agent, solution_agent, writer_agent],
    tasks=[rfp_task, solution_task, proposal_task],
    verbose=True
)

# -----------------------------
# 7. Input RFP (VERY IMPORTANT 🔥)
# -----------------------------
rfp_input = """
Client: ABC Corp

We are looking for a cloud-based learning management system (LMS)
that supports:
- AI-based personalized learning
- Real-time analytics dashboard
- Integration with existing HR systems
- Budget: $50,000
- Evaluation criteria: scalability, ease of use, and support
"""

# -----------------------------
# 8. Run Workflow
# -----------------------------
result = crew.kickoff(inputs={"input": rfp_input})

print("\n✅ FINAL PROPOSAL:\n")
print(result)
```
